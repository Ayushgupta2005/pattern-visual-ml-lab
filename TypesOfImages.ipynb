```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image /  Loads in BGR
img = cv2.imread("sample.jpg")   # put any image in same folder
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image (8-bit)")
plt.axis("off")
plt.show()
```

```python
# Thresholding
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

plt.imshow(binary, cmap='gray')
plt.title("Binary Image (0 or 1)")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(10,4))

plt.subplot(1,3,1)
plt.imshow(img)
plt.title("Color Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(binary, cmap='gray')
plt.title("Binary Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Load image
img = cv2.imread("sample.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img)
plt.title("Original Color Image (RGB)")
plt.axis("off")
plt.show()
```

```python
R, G, B = cv2.split(img)
```

```python
plt.figure(figsize=(10,6))

plt.subplot(2,2,1)
plt.imshow(R, cmap='gray')
plt.title("Red Channel")
plt.axis("off")

plt.subplot(2,2,2)
plt.imshow(G, cmap='gray')
plt.title("Green Channel")
plt.axis("off")

plt.subplot(2,2,3)
plt.imshow(B, cmap='gray')
plt.title("Blue Channel")
plt.axis("off")

plt.subplot(2,2,4)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.show()
```

```python
import numpy as np

zero = np.zeros_like(R)

red_only   = cv2.merge([R, zero, zero])
green_only = cv2.merge([zero, G, zero])
blue_only  = cv2.merge([zero, zero, B])

plt.figure(figsize=(10,6))

plt.subplot(2,2,1)
plt.imshow(red_only)
plt.title("Red Only Image")
plt.axis("off")

plt.subplot(2,2,2)
plt.imshow(green_only)
plt.title("Green Only Image")
plt.axis("off")

plt.subplot(2,2,3)
plt.imshow(blue_only)
plt.title("Blue Only Image")
plt.axis("off")

plt.subplot(2,2,4)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.show()
```
