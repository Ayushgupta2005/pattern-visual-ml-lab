# Experiment 5: Image Segmentation

```python
# Import libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image
img = cv2.imread("image.jpeg")

# Convert BGR image to RGB
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display original image
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.show()
```

## Converting Image to Grayscale

```python
# Convert RGB image to grayscale
grayscale_img = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

# Display grayscale image
plt.imshow(grayscale_img, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")

plt.show()
```

## Applying Threshold Segmentation

```python
# Apply binary thresholding
_, binary = cv2.threshold(grayscale_img, 127, 255, cv2.THRESH_BINARY)

# Display segmented image
plt.imshow(binary, cmap='gray')
plt.title("Binary Image")
plt.axis("off")
plt.show
```

## Applying Edge-Based Segmentation

```python
# Load image in grayscale format
img = cv2.imread('image.jpeg', cv2.IMREAD_GRAYSCALE)

# Check if image is loaded properly
assert img is not None, "File could not be read"

# Apply Canny Edge Detection
edges = cv2.Canny(img, 100, 200)

# Display edge-detected image
plt.imshow(edges, cmap='gray')
plt.title("Edges after Canny Function")
plt.axis("off")

plt.show()
```

## Applying Region-Based Segmentation

```python
# Import libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load grayscale image
img = cv2.imread('image.jpeg', cv2.IMREAD_GRAYSCALE)

# Apply Otsu Thresholding
_, binary = cv2.threshold(
    img,
    0,
    255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

# Create kernel for morphology
kernel = np.ones((5,5), np.uint8)

# Apply morphological opening
binary = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)

# Apply distance transform
dist = cv2.distanceTransform(binary, cv2.DIST_L2, 5)

# Threshold for segmentation
_, segmented = cv2.threshold(
    dist,
    0.3 * dist.max(),
    255,
    0
)

# Plot results
plt.subplot(1,2,1)
plt.title("Distance")
plt.imshow(dist, cmap='gray')

plt.subplot(1,2,2)
plt.title("Segmented")
plt.imshow(segmented, cmap='gray')

plt.show()
```

## Result

Image Segmentation techniques were successfully implemented using OpenCV.  
Thresholding, edge-based segmentation, and region-based segmentation were applied to segment the image.

## Conclusion

This experiment demonstrated different image segmentation techniques used to partition images into meaningful regions for visual analysis.
