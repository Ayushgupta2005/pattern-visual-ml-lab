## Dataset Description

The dataset used in this experiment is **Ice Cream Selling Data**.  
It contains two features:

- **Temperature (°C)** → Independent Variable  
- **Ice Cream Sales (units)** → Dependent Variable  

The dataset is used to analyze the non-linear relationship between temperature and ice cream sales. As temperature increases, ice cream sales generally increase in a curved pattern, making it suitable for non-linear regression analysis.

```python
# Importing numpy for numerical computations
import numpy as np

# Importing pandas for data handling and analysis
import pandas as pd

# Importing matplotlib for data visualization
import matplotlib.pyplot as plt
```

```python
# Loading the dataset from CSV file
df = pd.read_csv("Ice_cream_selling_data.csv")

# Displaying the first 5 rows of the dataset
df.head()
```

```python
# Displaying basic information about the dataset
df.info()
```

```python
# Checking dataset shape
print("Dataset Shape:", df.shape)
```

```python
# Extracting input and output variables
X = df["Temperature (°C)"].values
y = df["Ice Cream Sales (units)"].values
```

```python
# Visualizing the dataset using scatter plot
plt.scatter(X, y, color='blue')

# Labeling x-axis
plt.xlabel("Temperature (°C)")

# Labeling y-axis
plt.ylabel("Ice Cream Sales (units)")

# Adding title to the plot
plt.title("Temperature vs Ice Cream Sales")

# Displaying the plot
plt.show()
```

## Observation

From the scatter plot, it can be observed that the relationship between temperature and ice cream sales is non-linear in nature. Therefore, non-linear regression is more suitable for fitting the data compared to simple linear regression.

```python
# Defining a non-linear polynomial function
# y = ax^2 + bx + c

def predict(X, a, b, c):
    return a*(X**2) + b*X + c
```

```python
# Initializing model parameters
a = 0.1
b = 0.1
c = 0.1

# Setting learning rate
learning_rate = 0.001

# Number of training iterations
epochs = 1000

# Number of samples
n = len(X)
```

```python
# Training the non-linear regression model using Gradient Descent

for epoch in range(epochs):

    # Predicting output values
    y_pred = predict(X, a, b, c)

    # Calculating error
    error = y_pred - y

    # Calculating gradients
    da = (2/n) * np.sum(error * (X**2))
    db = (2/n) * np.sum(error * X)
    dc = (2/n) * np.sum(error)

    # Updating parameters
    a = a - learning_rate * da
    b = b - learning_rate * db
    c = c - learning_rate * dc
```

```python
# Displaying trained model parameters
print("Coefficient a:", a)
print("Coefficient b:", b)
print("Coefficient c:", c)
```

```python
# Predicting values using trained model
y_pred = predict(X, a, b, c)
```

```python
# Calculating Mean Squared Error (MSE)
mse = np.mean((y - y_pred)**2)

# Calculating Mean Absolute Error (MAE)
mae = np.mean(np.abs(y - y_pred))

# Displaying error metrics
print("Mean Squared Error (MSE):", mse)
print("Mean Absolute Error (MAE):", mae)
```

```python
# Sorting values for smooth curve plotting
sorted_idx = np.argsort(X)

X_sorted = X[sorted_idx]
y_sorted = y_pred[sorted_idx]
```

```python
# Plotting original data points
plt.scatter(X, y, color='blue', label='Original Data')

# Plotting fitted non-linear regression curve
plt.plot(X_sorted, y_sorted, color='red', linewidth=3, label='Fitted Curve')

# Labeling x-axis
plt.xlabel("Temperature (°C)")

# Labeling y-axis
plt.ylabel("Ice Cream Sales (units)")

# Adding plot title
plt.title("Non-Linear Regression Curve Fitting")

# Displaying legend
plt.legend()

# Showing the plot
plt.show()
```

## Conclusion

In this experiment, non-linear regression was implemented from scratch using Gradient Descent on the Ice Cream Selling dataset. A polynomial curve was fitted between temperature and ice cream sales. The fitted curve successfully captured the non-linear relationship present in the dataset. Error metrics such as MSE and MAE were also calculated to evaluate the model performance.
